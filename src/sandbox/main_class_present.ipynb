# Cross Check with the Mainfest data
Against the manifest to see whether `main_class_present == False` lines up with exit_code_1

```python
import pandas as pd

L = pd.read_csv("launchable_files.csv")
M = pd.read_csv("data/manifest.csv")

df = L.merge(M[["filename", "exit_reason", "opcode_count"]], on="filename")

print(pd.crosstab(df.main_class_present, df.exit_reason))
```
